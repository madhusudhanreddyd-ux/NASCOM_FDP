```python
import numpy as np #import numpy library for numerical calculation
```

```python
u=np.array([2,3]) #array u created
```

```python
v=np.array([5,6]) #array v created
print("u+v",u+v)
print("3*v",3*v) #perform addtion of arrays and multipy with constant
```

```python
d1=np.dot(u,v) #perform dot product between u and v
d2=u@v  #multiplication of u and v
cos_theta=np.dot(u,v)/(np.linalg.norm(u)*np.linalg.norm(v)) #calculate find angle between u and c
print("d1",d1)
print("d2",d2)
print("cos theta",cos_theta) #print theta value
```

```python
theta=np.arccos(cos_theta)
print(f"Angle: {np.degrees(theta):.1f}o") #calculate find angle between u and c
```

```python
w=np.array([3,-5])
```

```python
print("l1",np.linalg.norm(w,1))
print("l2",np.linalg.norm(w,2))
print("linf",np.linalg.norm(w,np.inf)) #find and prints three different vector norms
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
a,b=np.array([[1,2,3]]),np.array([[2,4,2]])
print(cosine_similarity(a,b)) #find cosine similarity of two arrays
```

```python
a=np.array([[1,3,1],[3,9,1],[1,4,3]])
b=np.array([3,9,1])

x=np.linalg.solve(a,b)  #solve two mathematical expressions
print(x) #dislay solve value
```
